## Part 1: Get a feel for the audio files

To begin, take a look at the data you'll use during this modeling cycle. You'll find two files located in the `hub_data_share/data/` directory, `guitar-mono.wav` and `banjo-mono-wav`. These are audio recordings of each instrument plucked at a musical C note.

&#x2705; Read the files into the jupyter notebook using [**scipy.io.wavfile.read**](https://docs.scipy.org/doc/scipy/reference/generated/scipy.io.wavfile.read.html).

&#x2705; Use [**IPython.display.Audio**](https://ipython.readthedocs.io/en/stable/api/generated/IPython.display.html) to play the audio out loud from your jupyter notebook.

&#x2705; Create a visualization of each audio recording. Discuss how they compare qualitatively.

```python
from scipy.io import wavfile
import numpy as np
import IPython.display
from IPython.display import Audio
import matplotlib.pyplot as plt

banjo_sample_rate, banjo_data = wavfile.read('/home/ryancole/hub_data_share/data/banjo-mono.wav')
guitar_sample_rate, guitar_data = wavfile.read('/home/ryancole/hub_data_share/data/guitar-mono.wav')

Audio(banjo_data, rate=banjo_sample_rate)
Audio(guitar_data, rate=guitar_sample_rate)

duration = guitar_data.shape[0] / guitar_sample_rate
time = np.linspace(0., duration, guitar_data.shape[0])
plt.plot(time, guitar_data, label='guitar')

duration = banjo_data.shape[0] / banjo_sample_rate
time = np.linspace(0., duration, banjo_data.shape[0])
plt.plot(time, banjo_data, label='banjo')
plt.legend()
plt.xlabel("Time [s]")
plt.ylabel("Amplitude")
plt.show()
```
